### Intervention Steering - Real Data

```python
data = "real"
steering_mode = "intervention"

## choose one of the following
# synthetic trade, intervention steering, agressive
steering = "agressive"
# synthetic trade, intervention steering, neutral
steering = "neutral"
# synthetic trade, intervention steering, agreeable
steering = "agreeable"

gpt_model = configs['gpt_model']
trade_environment = configs['trade_environment']
trade_config = trade_configs[data]
trade = trade_config['trade']
money = trade_config['money']
currency = trade_config['currency']
cost_of_production = trade_config['cost_of_production']
willingness_to_pay = trade_config['willingness_to_pay']
product_description = trade_config['product_description']
currency_description = trade_config['currency_description']
steering_config = steering_configs[steering_mode][steering]
steering_messages = steering_config['steering_message']

trial_number = 1 # 1, 2, 3, 4, 5
log_name = f"intervention_real_{trial_number}"
```

### Template

```python
try:
    a1 = ChatGPTAgent(agent_name=AGENT_ONE, model = gpt_model)
    a2 = HumanAgent(agent_name=AGENT_TWO)

    c = BuySellGame(
        players=[a1, a2],
        iterations=10,
        player_goals=[
            SellerGoal(cost_of_production=Valuation({trade: cost_of_production})),
            BuyerGoal(willingness_to_pay=Valuation({trade: willingness_to_pay})),
        ],
        player_starting_resources=[
            Resources({trade: 1}),
            Resources({MONEY_TOKEN: money}),
        ],
        player_conversation_roles=[
            f"You are {AGENT_ONE}.",
            f"You are {AGENT_TWO}.",
        ],

        # new social behaviors
        player_social_behaviour=[
            steering_messages if steering_mode == "initialization" else "", # only for initialization steering mode
            ""
        ],

        player_trade_environment=[
            trade_environment + "\n" + (currency_description if currency == "USD" else ""),  # currency description is only for real data
            ""
        ],
        log_dir=f"../example_logs/buysell/{log_name}",
    )

    c.run()
except Exception as e:
    exception_type = type(e).__name__
    exception_message = str(e)
    stack_trace = traceback.format_exc()

    # Print or use the information as needed
    print(f"Exception Type: {exception_type}")
    print(f"Exception Message: {exception_message}")
    print(f"Stack Trace:\n{stack_trace}")
```
